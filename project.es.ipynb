#### Ejercicio 21. Realiza un trazado de las coordenadas (columnas latitud y longitud) del cinturón sur de Madrid por color de cada población (debes usar el subconjunto obtenido del Ejercicio 14) (★★★★)

Ejecuta la siguiente celda y luego comienza a codear en la siguiente. Debes implementar un código simple que transforme las columnas de coordenadas en un diccionario de Python (agrega más información si es necesario) y agrégala al mapa.

```python
from ipyleaflet import Map, basemaps

# Mapa centrado en Madrid
map = Map(center = (40.35, -3.8), zoom = 11.5, min_zoom = 1, max_zoom = 20, 
    basemap=basemaps.OpenStreetMap.Mapnik)
map
```

```python
## Aquí: traza la coordenadas de los estados

## PON TU CÓDIGO AQUÍ:
from ipyleaflet import Marker, AwesomeIcon

cinturon_sur['latitude'] = cinturon_sur['latitude'].astype(str).str.replace(',', '.', regex=False).astype(float)
cinturon_sur['longitude'] = cinturon_sur['longitude'].astype(str).str.replace(',', '.', regex=False).astype(float)

colores_disponibles = ['red',  'green', 'orange', 'purple', 'pink', 'darkred', 'darkgreen', 'darkblue']
poblaciones = cinturon_sur['level5'].unique()
colores_poblacion = dict(zip(poblaciones, colores_disponibles[:len(poblaciones)]))

for poblacion, color in colores_poblacion.items():
    viviendas = cinturon_sur[cinturon_sur['level5'] == poblacion]
    for _, row in viviendas.iterrows():
        icon = AwesomeIcon(marker_color=color, icon='circle', icon_color='white')
        marker = Marker(location=(row['latitude'], row['longitude']), icon=icon)
        map.add_layer(marker)

map
```
